```python
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
import pandas as pd
import datetime
import re

이닝=[]
투수=[]
타자=[]
투구=[]
타석결과=[]
이전상황=[]
이후상황=[]
홈팀=[]
원정팀=[]
날짜=[]
구장=[]
LEV=[]
REs=[]
REa=[]
WPe=[]
WPa=[]
경기결과=[]

 
driver=webdriver.Chrome()
for i in range(0, 500):  
    game_no = "2023" + str(i).zfill(4)
    url = f"https://statiz.sporki.com/schedule/?m=gamelogs&s_no={game_no}"
    driver.get(url)
    time.sleep(10)
    html=driver.page_source
    soup=BeautifulSoup(html,'html.parser')
    try:
        addition=soup.select('.table_type03')[3]
        consumer=addition.select("div.table_type03 table tbody tr:nth-of-type(2) td.align_left")
        text = consumer[0].get_text(strip=True)  
        number= re.sub(r'\D', '', text) 
        if int(number)>500:
            home=soup.select('.box_head')[0].get_text(strip=True)
            away=soup.select('.box_head')[1].get_text(strip=True)
            stadium=soup.select('.txt')[1].text
            homescore = soup.select_one('.score .num span:nth-of-type(1)').get_text(strip=True)
            awayscore = soup.select_one('.score .num span:nth-of-type(2)').get_text(strip=True)
            table=soup.select('.table_type03')[2]
            tbody = table.find('table').find('tbody')
            rows = tbody.find_all('tr')
            for row in rows:
                cols = row.find_all('td')
                이닝.append(cols[0].get_text(strip=True))
                투수.append(cols[1].get_text(strip=True))
                타자.append(cols[2].get_text(strip=True))
                투구.append(cols[3].get_text(strip=True))
                타석결과.append(cols[4].get_text(strip=True))
                이전상황.append(cols[5].get_text(strip=True))
                이후상황.append(cols[6].get_text(strip=True))
                LEV.append(cols[7].get_text(strip=True))
                REs.append(cols[8].get_text(strip=True))
                REa.append(cols[9].get_text(strip=True))
                WPe.append(cols[10].get_text(strip=True))
                WPa.append(cols[11].get_text(strip=True))
                홈팀.append(home.split(',')[0].strip())
                원정팀.append(away.split(',')[0].strip())
                날짜.append(stadium.split(',')[0].strip())
                구장.append(stadium.split(',')[1].strip())
                경기결과.append(awayscore+':'+homescore)
    except Exception:
        continue



df = pd.DataFrame({
    '구장':구장,
    '날짜':날짜,
    '홈팀':홈팀,
    '원정팀':원정팀,
    '경기결과':경기결과,
    '이닝': 이닝,
    '투수': 투수,
    '타자': 타자,
    '투구': 투구,
    '타석결과': 타석결과,
    '이전상황': 이전상황,
    '이후상황': 이후상황,
    'LEV': LEV,
    'REs': REs,
    'REa': REa,
    'WPe': WPe,
    'WPa': WPa
})

df.to_csv("gamelog23-1.csv", index=False, encoding="utf-8-sig")
```
